```python
# Helper function to convert DMS to Degrees
import re

def dms_to_decimal(dms):
    # Extract numbers
    parts = re.findall(r"\d+\.?\d*", dms)
    degrees, minutes, seconds = map(float, parts)

    decimal = degrees + minutes / 60 + seconds / 3600

    # Handle direction
    if any(x in dms for x in ["S", "W"]):
        decimal = -decimal

    return decimal
```

```python
# Apply DMS to Decimal conversion
Hong_Kong_River_Water_Quality_coordinates_df["Latitude"] = Hong_Kong_River_Water_Quality_coordinates_df["Latitude"].apply(dms_to_decimal)
Hong_Kong_River_Water_Quality_coordinates_df["Longitude"] = Hong_Kong_River_Water_Quality_coordinates_df["Longitude"].apply(dms_to_decimal)
```

```python
import numpy as np
import pandas as pd

# Helper function to clean numeric values
def clean_numeric(val):
    if pd.isna(val):
        return np.nan
    
    val = str(val).strip().upper()
    
    if val in ["ND", "LOD", "BDL", "-", "", "NA"]:
        return np.nan
    
    if val.startswith("<"):
        try:
            return float(val.replace("<", "")) / 2
        except:
            return np.nan
    
    val = val.replace(",", "")
    
    try:
        return float(val)
    except:
        return np.nan
```

```python
for col in Hong_Kong_River_Water_Quality_merged_df.columns:
    if Hong_Kong_River_Water_Quality_merged_df[col].dtype == "object" and col not in ["River", "Date"]:
        Hong_Kong_River_Water_Quality_merged_df[col] = Hong_Kong_River_Water_Quality_merged_df[col].apply(clean_numeric)
```

```python
# µg -> mg
England_River_Water_Quality_merged_df["TP_mgL"] = England_River_Water_Quality_merged_df["TP_ugL"] / 1000

# NO2 -> N
England_River_Water_Quality_merged_df["NO2_N_mgL"] = England_River_Water_Quality_merged_df["NO2_mgL"] * 0.304

# NO3 -> N
England_River_Water_Quality_merged_df["NO3_N_mgL"] = England_River_Water_Quality_merged_df["NO3_mgL"] * 0.226

ug_cols = ["Fe_ugL", "Mn_ugL", "Zn_ugL", "Cu_ugL", "Chl_a_ugL", "TP_ugL"]

for col in ug_cols:
    England_River_Water_Quality_merged_df[col.replace("_ugL", "_mgL")] = England_River_Water_Quality_merged_df[col] / 1000
```

```python
England_River_Water_Quality_merged_df = England_River_Water_Quality_merged_df.drop(columns=ug_cols)
England_River_Water_Quality_merged_df = England_River_Water_Quality_merged_df.drop(columns=["NO2_mgL", "NO3_mgL"])
England_River_Water_Quality_merged_df.info()
```

```python
Combined_River_Water_Quality_df = pd.concat([Hong_Kong_River_Water_Quality_merged_df, England_River_Water_Quality_merged_df], ignore_index=True)
```

```python
Combined_River_Water_Quality_df.info()
```

```python
# To filter data based on 'Date' column from 2000 to 2025
Combined_River_Water_Quality_df['Date'] = pd.to_datetime(Combined_River_Water_Quality_df['Date'], dayfirst=True, errors='coerce')
Combined_River_Water_Quality_df = Combined_River_Water_Quality_df[(Combined_River_Water_Quality_df['Date'] >= '2000-01-01') & (Combined_River_Water_Quality_df['Date'] <= '2025-12-31')]
Combined_River_Water_Quality_df = Combined_River_Water_Quality_df.reset_index(drop=True)
```
